# 09 — Repeated-seed and estimation-window robustness

Repeat the rolling experiment so conclusions do not depend on one random seed. The first run uses classical models and is relatively quick. A separate cell adds DI-VAR because retraining it for each seed is intentionally more expensive.

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
CACHE = ROOT / 'results/notebook_cache'
CACHE.mkdir(parents=True, exist_ok=True)
from innovcal.experiments.repeated import (
    run_repeated_rolling_experiment,
    summarize_repeated_results,
)
```

```python
returns = pd.read_csv(ROOT / 'data/processed/financial_returns.csv', index_col=0).to_numpy()
SEEDS = (101, 202, 303)
CLASSICAL = ('gaussian', 'student_t', 'bootstrap', 'block_bootstrap', 'volatility_bootstrap')

classical_repeats = run_repeated_rolling_experiment(
    returns,
    seeds=SEEDS,
    methods=CLASSICAL,
    window_lengths=(None, 1000),
    horizon=20,
    origin_step=20,
    n_paths=250,
)
display(summarize_repeated_results(classical_repeats))
```

## Optional DI-VAR repeats

Run this cell for the final analysis. Validation-based early stopping selects a checkpoint separately for each seed. Start with the expanding window; add the 1000-day window only if that comparison is central to the written claim.

```python
RUN_DIFFUSION_REPEATS = False
diffusion_repeats = pd.DataFrame()
if RUN_DIFFUSION_REPEATS:
    diffusion_repeats = run_repeated_rolling_experiment(
        returns,
        seeds=SEEDS,
        methods=('diffusion',),
        window_lengths=(None,),
        horizon=20,
        origin_step=20,
        n_paths=250,
        diffusion_options={
            'timesteps': 100,
            'epochs': 300,
            'hidden_dim': 128,
            'batch_size': 64,
            'validation_fraction': 0.2,
            'early_stopping_patience': 40,
            'verbose': False,
        },
    )
    display(summarize_repeated_results(diffusion_repeats))
else:
    print('Set RUN_DIFFUSION_REPEATS=True for final DI-VAR repetitions.')
```

## Optional CDI-VAR repeats

This is the final, most expensive comparison. It repeats training, conditional recursive sampling, and explicit calibration for every seed. Start with three seeds and the expanding window so it remains directly comparable with the DI-VAR repetitions.

```python
RUN_CDI_VAR_REPEATS = True
cdi_var_repeats = pd.DataFrame()
if RUN_CDI_VAR_REPEATS:
    cdi_var_repeats = run_repeated_rolling_experiment(
        returns,
        seeds=SEEDS,
        methods=('cdi_var',),
        window_lengths=(None,),
        horizon=20,
        origin_step=20,
        n_paths=250,
        cdi_var_options={
            'context_lags': 5,
            'volatility_span': 60,
            'validation_fraction': 0.15,
            'calibration_fraction': 0.15,
            'calibration_horizons': (1, 5, 20),
            'calibration_paths': 32,
            'calibration_shrinkage': 0.5,
            'calibration_bounds': (0.8, 1.25),
            'adaptive_calibration_window': 12,
            'timesteps': 50,
            'epochs': 300,
            'hidden_dim': 128,
            'time_embedding_dim': 32,
            'batch_size': 64,
            'early_stopping_patience': 40,
            'verbose': False,
        },
    )
    display(summarize_repeated_results(cdi_var_repeats))
else:
    print('Set RUN_CDI_VAR_REPEATS=True for final CDI-VAR repetitions.')
```

```python
repeated_results = pd.concat([classical_repeats, diffusion_repeats, cdi_var_repeats], ignore_index=True)
repeated_results.to_csv(CACHE / 'repeated_rolling_results.csv', index=False)
summary = summarize_repeated_results(repeated_results)
summary.to_csv(CACHE / 'repeated_rolling_summary.csv', index=False)
display(summary)
```
