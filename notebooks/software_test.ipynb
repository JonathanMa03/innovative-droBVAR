```python
import numpy as np
import pandas as pd

# ---------------------------------------------------------------------
# Tiny smoke test for metrics.py
# ---------------------------------------------------------------------
rng = np.random.default_rng(123)

forecast_draws = rng.normal(size=(500, 4, 2))
summary = summarize_forecast_draws(forecast_draws, alpha=0.05)

print("Summary shapes:")
print("mean :", summary["mean"].shape)
print("lower:", summary["lower"].shape)
print("upper:", summary["upper"].shape)

y_true = np.array([1.0, 2.0, 3.0])
y_pred = np.array([1.1, 1.9, 2.8])

print("\nRMSE:", rmse(y_true, y_pred))
print("MAE :", mae(y_true, y_pred))

lower = np.array([0.9, 1.7, 2.7])
upper = np.array([1.2, 2.3, 3.2])
print("Coverage:", predictive_interval_coverage(y_true, lower, upper))

X_emp = rng.normal(size=(300, 3))
X_gen = rng.normal(loc=0.1, scale=1.1, size=(300, 3))

metrics_out, emp_sum, gen_sum, emp_cov, gen_cov = compare_distribution_summaries(
    X_generated=X_gen,
    X_empirical=X_emp,
    columns=["x1", "x2", "x3"]
)

print("\nDistribution comparison metrics:")
print(metrics_out)

print("\nEmpirical summary:")
display(emp_sum.round(4))

print("Generated summary:")
display(gen_sum.round(4))

row = final_horizon_summary_row(
    model_name="TestModel",
    summary_obj=summary,
    horizon=4,
    variable_names=["y1", "y2"],
    variable_index=0,
)
print("\nFinal horizon row:")
print(row)

df_all = final_horizon_summary_all_variables(
    model_name="TestModel",
    summary_obj=summary,
    horizon=4,
    variable_names=["y1", "y2"],
)
print("\nFinal horizon all variables:")
display(df_all.round(4))
```
