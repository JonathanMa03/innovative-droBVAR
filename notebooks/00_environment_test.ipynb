```python
# Imports
import sys
from pathlib import Path

ROOT = Path.cwd().parent
sys.path.append(str(ROOT))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy
import torch

from src.utils.seeds import set_seed
from src.utils.device import get_device
from src.utils.logging_utils import (
    ensure_dir,
    make_run_dir,
    save_metrics_json,
    save_array_npy,
)

from src.data.loaders import (
    save_dataframe,
    load_dataframe,
    save_array,
    load_array,
)
```

```python
# Package versions
print("Python:", sys.version)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("SciPy:", scipy.__version__)
print("PyTorch:", torch.__version__)
```

```python
# Torch Device
device = get_device()

print("Selected device:", device)

if torch.cuda.is_available():
    print("CUDA available")

if hasattr(torch.backends, "mps"):
    print("MPS available:", torch.backends.mps.is_available())
```

```python
# Seeding
set_seed(123)

x1 = np.random.normal(size=5)

set_seed(123)

x2 = np.random.normal(size=5)

print("x1:", x1)
print("x2:", x2)

print("Reproducible:", np.allclose(x1, x2))
```

```python
# Paths
data_dir = ROOT / "data"
results_dir = ROOT / "results"
logs_dir = ROOT / "logs"

print("Data dir exists:", data_dir.exists())
print("Results dir exists:", results_dir.exists())
print("Logs dir exists:", logs_dir.exists())
```

```python
# Create Experiment Directory
run_dir = make_run_dir(
    base_dir=logs_dir,
    run_name="environment_test",
)

print(run_dir)
```

```python
# Save/Load Tests
test_array = np.random.normal(size=(10, 3))

array_path = run_dir / "test_array.npy"

save_array_npy(
    test_array,
    array_path,
)

loaded_array = load_array(array_path)

print("Saved shape:", test_array.shape)
print("Loaded shape:", loaded_array.shape)

print(
    "Arrays identical:",
    np.allclose(test_array, loaded_array),
)
```

```python
df = pd.DataFrame({
    "a": np.random.normal(size=5),
    "b": np.random.normal(size=5),
})

csv_path = run_dir / "test_dataframe.csv"

save_dataframe(
    df,
    csv_path,
)

loaded_df = load_dataframe(csv_path)

display(loaded_df)
```

```python
metrics = {
    "rmse": 1.23,
    "coverage": 0.91,
    "ece": 0.04,
}

metrics_path = run_dir / "metrics.json"

save_metrics_json(
    metrics,
    metrics_path,
)

print(metrics_path)
```
